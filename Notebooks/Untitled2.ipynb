**fit():** Analiza los datos y calcula los parámetros necesarios para realizar la normalización, como los valores mínimos y máximos.

**transform():** Aplica la normalización a los datos utilizando los parámetros calculados por fit().

**fit_transform():** Combina ambas funciones, analizando y transformando los datos en un solo paso.
